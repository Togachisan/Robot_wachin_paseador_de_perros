```python
import random
import time

# ==========================================
# DEFINICIÓN DE ACTUADORES (Salidas del Robot)
# ==========================================
def avanzar(velocidad):
    print(f"🚀 [AVANZAR] Moviendo ruedas a {velocidad} m/s...")

def frenar():
    print("🛑 [FRENAR] Deteniendo el movimiento de forma gradual.")

def girar(grados):
    print(f"🔄 [GIRAR] Modificando trayectoria {grados} grados para esquivar obstáculo.")

def emitir_sonido(tipo_sonido):
    sonidos = {
        "alerta_espera": "🔊 *Beep-beep* Nivel bajo o perro no listo. Esperando...",
        "esperar_luz_verde": "🔊 *Ding* Esperando cambio de semáforo a verde...",
        "esperar_perro": "🔊 *Bip* Tensión alta en la correa. Pausando marcha...",
        "paseo_completado": "🎶 *Tatam!* ¡Paseo completado con éxito!"
    }
    print(f"[{sonidos.get(tipo_sonido, '🔊 Emitiendo sonido...')}]")

def conectar_correa():
    print("🔗 [BRAZO MECÁNICO] Conectando y asegurando el collar del can con seguridad.")

def desconectar_correa():
    print("🔓 [BRAZO MECÁNICO] Desconectando la correa de forma segura.")

def activar_modo_proteccion():
    print("🛡️ [MODO PROTECCIÓN] Cambiando al lado opuesto de la acera por peatón cercano.")

def desplegar_eco_scoop():
    print("🧹 [ECO-SCOOP] Desplegando pala recolectora y bolsa biodegradable...")
    print("✨ Necesidades gestionadas y recogidas limpiamente.")

def activar_rodillo_limpieza():
    print("🧼 [RODILLO LIMPIEZA] Activando sistema de microfibra húmeda para limpiar las patas.")

def dispensar_premio():
    print("🦴 [DISPENSADOR] ¡Premio comestible liberado para el can!")

def abrir_puerta_inteligente():
    print("🚪 [CASA] Abriendo puerta inteligente para iniciar la salida...")

def invertir_ruta_gps():
    print("🛰️ [GPS] Ruta invertida. Iniciando recorrido de regreso a casa...")


# ==========================================
# SIMULACIÓN PRINCIPAL: PaseoRobotico
# ==========================================
def iniciar_paseo_robotico():
    print("=== INICIO DEL SISTEMA: PaseoRobotico ===")

    # ------------------------------------------
    # FASE 1: PREPARACIÓN
    # ------------------------------------------
    print("\n--- FASE 1: PREPARACIÓN ---")
    # Simulamos lectura de sensores iniciales (puedes ajustar o usar random)
    bateria_nivel = random.randint(75, 100)
    reconocimiento_facial_canino = random.choice(["verde", "alerta"])

    print(f"🔋 Batería actual: {bateria_nivel}%")
    print(f"🐶 Reconocimiento facial canino: '{reconocimiento_facial_canino}'")
    time.sleep(1)

    if bateria_nivel > 80 and reconocimiento_facial_canino == "verde":
        conectar_correa()
    else:
        emitir_sonido("alerta_espera")
        print("❌ El robot no puede iniciar el paseo. Fin del proceso.")
        return  # Termina la ejecución si no se cumplen las condiciones

    time.sleep(1.5)

    # ------------------------------------------
    # FASE 2: SALIDA
    # ------------------------------------------
    print("\n--- FASE 2: SALIDA ---")
    abrir_puerta_inteligente()

    # Simulación del LIDAR al salir
    distancia_obstaculo_lidar = random.randint(40, 90)
    print(f"📡 LIDAR inicial - Distancia a obstáculo: {distancia_obstaculo_lidar} cm")

    while distancia_obstaculo_lidar > 50:
        avanzar(0.5)
        time.sleep(1)
        # Actualizamos valor para simular avance en bucle
        distancia_obstaculo_lidar = random.randint(30, 80)
        print(f"   ↳ Distancia actual: {distancia_obstaculo_lidar} cm")

    print("🚪 Pasillo despejado o umbral superado.")
    time.sleep(1.5)

    # ------------------------------------------
    # FASES 3 Y 4: BUCLE DE PASEO ACTIVO Y PIT-STOP
    # ------------------------------------------
    print("\n--- FASE 3 Y 4: PASEO ACTIVO ---")
    tiempo_paseo_transcurrido = 0
    limite_tiempo = 5  # Reducido a 5 iteraciones en la simulación para agilizar Colab

    while tiempo_paseo_transcurrido < limite_tiempo and bateria_nivel > 20:
        tiempo_paseo_transcurrido += 1
        bateria_nivel -= random.randint(2, 5) # El robot gasta batería
        print(f"\n⏱️ Minuto/Ciclo {tiempo_paseo_transcurrido} de {limite_tiempo} | 🔋 Batería: {bateria_nivel}%")

        # 1. Control de Cruces y Semáforos
        cruce_peatonal_detectado = random.choice([True, False])
        if cruce_peatonal_detectado:
            estado_semaforo = random.choice(["rojo", "verde"])
            print(f"🚦 Cruce peatonal detectado. Semáforo en: {estado_semaforo.upper()}")

            if estado_semaforo == "rojo":
                frenar()
                emitir_sonido("esperar_luz_verde")
                # Bucle de espera en rojo
                while estado_semaforo == "rojo":
                    frenar()
                    time.sleep(1)
                    # Simulamos si cambia a verde en el siguiente chequeo
                    estado_semaforo = random.choice(["rojo", "verde"])
                    print(f"   ⏳ Esperando... Semáforo ahora está: {estado_semaforo.upper()}")
            elif estado_semaforo == "verde":
                avanzar(0.4)

        # 2. Control de Tensión de Correa
        tension_correa = random.choice(["baja", "normal", "alta"])
        print(f"🔗 Tensión de correa sensada: '{tension_correa}'")
        if tension_correa == "alta":
            frenar()
            emitir_sonido("esperar_perro")
        elif tension_correa == "baja":
            avanzar(0.3)
        else:
            avanzar(0.8)

        # 3. Control de Obstáculos Dinámicos (LIDAR)
        distancia_obstaculo_lidar = random.randint(80, 120)
        if distancia_obstaculo_lidar < 100:
            activar_modo_proteccion()
            girar(90)

        # 4. Gestión de Necesidades (Pit-Stop)
        patron_olfateo_detectado = random.choice([True, False, False]) # Menos probabilidad para que no ocurra siempre
        if patron_olfateo_detectado:
            print("🐕 El perro está dando vueltas en círculos...")
            frenar()
            desplegar_eco_scoop()

        time.sleep(2) # Pausa entre ciclos de paseo

    # ------------------------------------------
    # FASE 5: RETORNO Y CIERRE
    # ------------------------------------------
    print("\n--- FASE 5: RETORNO Y CIERRE ---")
    invertir_ruta_gps()

    sensor_recibidor_casa = False
    print("🏠 Buscando señal de baliza en casa...")

    # Simulamos el acercamiento hasta encontrar la casa
    intentos_regreso = 0
    while not sensor_recibidor_casa and intentos_regreso < 3:
        avanzar(0.5)
        intentos_regreso += 1
        time.sleep(1)
        if intentos_regreso == 3:
            sensor_recibidor_casa = True # Forzamos la llegada en la simulación
            print("📡 ¡Señal recibida en casa detectada!")

    frenar()
    activar_rodillo_limpieza()
    desconectar_correa()
    dispensar_premio()
    emitir_sonido("paseo_completado")
    print("\n=== ¡MISIÓN CUMPLIDA! EL PERRO ESTÁ EN CASA Y FELIZ ===")

# Ejecutar la simulación
iniciar_paseo_robotico()
```
